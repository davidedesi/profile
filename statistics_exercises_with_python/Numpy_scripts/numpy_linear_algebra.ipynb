```python
# find the inverse of a 2x2 matrix 
# exercise 1

# imnport the Fraction class from the fractions module to display the results as fractions
from fractions import Fraction
np.set_printoptions(
    formatter={'all': lambda x: str(Fraction(x).limit_denominator())}
)

# create a 2x2 matrix
matrix_2x2 = np.array([[-9, -6], [0, 2]])
# compute the determinant of the 2x2 matrix
det_matrix_2x2 = np.linalg.det(matrix_2x2)
print("Determinant of the 2x2 matrix:", det_matrix_2x2)
# check if the determinant is not zero before computing the inverse
if det_matrix_2x2 != 0:
    inverse_matrix_2x2 = np.linalg.inv(matrix_2x2)
    print("Inverse of the 2x2 matrix:\n", inverse_matrix_2x2)
else:
    print("The 2x2 matrix is singular and does not have an inverse.")
```

```python
# create a program to solve the system of equations using the inverse of the matrix
# take as an input the coefficients of the variables and the constants of the equations, 
# then compute the inverse of the coefficient matrix and multiply it by the constants vector to find the solution.
# the program should be able to handle any number of equations and variables, as long as the coefficient matrix is square and invertible.

# import the Fraction class from the fractions module to display the results as fractions
from fractions import Fraction
np.set_printoptions(
    formatter={'all': lambda x: str(Fraction(x).limit_denominator())}
)

# set a random seed for reproducibility
seed = 42

# create random 2x2 matrix and 2x1 vector
n = 2
rng = np.random.default_rng(seed)

A = rng.integers(-10, 11, size=(n, n)).astype(float)
b = rng.integers(-10, 11, size=(n, 1)).astype(float)
print("Random 2x2 matrix A:\n", A)
print("Random 2x1 vector b:\n", b)

# compute the determinant of the coefficient matrix
det_A = np.linalg.det(A)
print("Determinant of the coefficient matrix A:", det_A)

if det_A != 0:
    # compute the inverse of the coefficient matrix
    A_inv = np.linalg.inv(A)
    print("Inverse of the coefficient matrix A:\n", A_inv)
    
    # compute the solution by multiplying the inverse of the coefficient matrix with the constants vector
    solution = A_inv @ b
    print("Solution of the system of equations:\n", solution)

```

```python
import numpy as np

def solve_random_systems(num_systems, n, seed, regenerate_singular):
    """
    Generate and solve `num_systems` random linear systems Ax = b
    where A is n x n and b is n x 1.

    Args:
      num_systems (int): how many systems to generate and solve.
      n (int): size of the square coefficient matrix.
      seed (int or None): RNG seed for reproducibility. None => random seed.
      regenerate_singular (bool): if True, regenerate A until invertible.
                                 if False, skip singular systems and report them.

    Returns:
      results (list of dict): each dict contains 'A', 'b', 'x', 'status', 'det' (if computed)
    """
    rng = np.random.default_rng(seed)
    results = []

    for i in range(num_systems):
        attempt = 0
        while True:
            attempt += 1
            # Example distribution: integers from -10..10
            A = rng.integers(-10, 11, size=(n, n)).astype(float)
            b = rng.integers(-10, 11, size=(n, 1)).astype(float)

            # determinant check (use isclose for float precision)
            detA = np.linalg.det(A)
            if not np.isclose(detA, 0.0):
                # solvable
                try:
                    x = np.linalg.solve(A, b)  # more stable than inv(A) @ b
                    results.append({'A': A, 'b': b, 'x': x, 'status': 'solved', 'det': detA})
                except np.linalg.LinAlgError as e:
                    # numerically unstable or singular despite det check
                    results.append({'A': A, 'b': b, 'x': None, 'status': f'error: {e}', 'det': detA})
                break
            else:
                # A is singular
                if regenerate_singular:
                    # try again (but avoid infinite loop: cap attempts)
                    if attempt >= 100:
                        results.append({'A': A, 'b': b, 'x': None, 'status': 'singular_after_retries', 'det': detA})
                        break
                    continue
                else:
                    results.append({'A': A, 'b': b, 'x': None, 'status': 'singular', 'det': detA})
                    break

    return results


if __name__ == "__main__":
    res = solve_random_systems(num_systems=2, n=2, seed=42, regenerate_singular=True)
    solved = sum(1 for r in res if r['status'] == 'solved')
    print(f"Solved {solved}/{len(res)} systems.\n")

    # Print first 5 solutions for inspection
    for i, r in enumerate(res[:5], start=1):
        print(f"--- System {i} ---")
        print("A =\n", r['A'])
        print("b =\n", r['b'].ravel())
        print("det(A) =", r['det'])
        print("x =\n", r['x'].ravel() if r['x'] is not None else None)
        print("status:", r['status'])
        print()
```

### Exercise: compute Eigendecomposition using Python numpy

Eigendecomposition is a matrix factorization method in which a square matrix is factored (decomposed) into three multiplicative matrices. Such factorization is called eigendecomposition because the entries of the matrices are eigenvalues and eigenvectors of the original square matrix. Mathematically, a square matrix A can be factored into three matrices such that: A = PDP<sup>-1</sup>

where P is the matrix of the eigenvectors of A, D is a diagonal matrix whose nonzero elements are the ordered eigenvalues of the matrix, and P<sup>-1</sup> is the inverse of the matrix P.

```python
import numpy as np


# create a 2x2 matrix
matrix_2x2 = np.array([[4 ,1], [2, 3]])

# step 1: Compute the eigenvalues and eigenvectors with NumPy
eigenvalues, eigenvectors = np.linalg.eig(matrix_2x2)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

# step 2: Extract the  eigenvector and the inverse of the eigenvector
V = eigenvectors
V_inverse = np.linalg.inv(V)

print("Inverse of Matrix (eigenvectors):\n", V_inverse)

#step 3: Verify the diagonalization property A = VDV⁻¹ 
D = np.diag(eigenvalues)

# Reconstruct the original matrix A using the diagonalization property
A_reconstructed = V @ D @ V_inverse
print("Diagonal matrix D:\n", D)
print("VDV⁻¹:\n", A_reconstructed)
print("Original matrix A:\n", matrix_2x2)
is_diagonalized = np.allclose(A_reconstructed, matrix_2x2)
print("Is A = VDV⁻¹?", is_diagonalized)
```

### Projection exercise

```python
# import the Fraction class from the fractions module to display the results as fractions
from fractions import Fraction
np.set_printoptions(
    formatter={'all': lambda x: str(Fraction(x).limit_denominator())}
)
# create a 2x2 matrix
a = np.array([[2,1],[1,2]])
a_transpose = a.T
print("a:\n", a)
print("Transpose of a:\n", a_transpose)

# compute aᵀ·a
b = a_transpose @ a
print("aᵀ·a:\n", b)

# compute the inverse of aᵀ·a
b_inverse = np.linalg.inv(b)
print("Inverse of aᵀ·a:\n", b_inverse)

# compute the projection matrix onto the column space of a using the formula: P = A(AᵀA)⁻¹Aᵀ
proj_a = a @ b_inverse @ a_transpose
print("Projection matrix onto the column space of a:\n", proj_a)

```

```python
# Use that projection matrix to project the vector [3, 1] onto the column space of a
vector = np.array([[3], [1]])
projected_vector = proj_a @ vector
print("Projection of the vector [3, 1] onto the column space of a:\n", projected_vector)
```
