### Exercise 1 – Basic matrix creation and shapes
1. Create the following matrices in NumPy:

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[1, 0], [0 ,1], [1, -1]])
c = np.array([[2, -1, 0], [0, 3, 1], [4, 1, -2]])
```

```python
# print the arrays
print("Array a:\n", a)  # Output: [[1 2 3] [4 5 6]]
print("Array b:\n", b)  # Output: [[1 0] [0 1] [1 -1]]
print("Array c:\n", c)  # Output: [[2 -1 0] [0 3 1] [4 1 -2]]
```

2. Verify their shape, ndim, and dtype using NumPy.

```python
# get the data type, number of dimensions and shape of array a
print("Data type of array a:", a.dtype)
print("Number of dimensions of array a:", a.ndim)
print("Shape of array a:", a.shape) 

# get the data type, number of dimensions and shape of array b
print("----------------------------------")
print("Data type of array b:", b.dtype)
print("Number of dimensions of array b:", b.ndim)
print("Shape of array b:", b.shape)

# get the data type, number of dimensions and shape of array c
print("----------------------------------")
print("Data type of array c:", c.dtype)
print("Number of dimensions of array c:", c.ndim)   
print("Shape of array c:", c.shape)
```

```python
a_transpose = a.T
print("array a:\n", a)  # Output: [[1 2 3] [4 5 6]]
print("Transpose of array a:\n", a_transpose)  # Output: [[1 4] [2 5] [3 6]]

b_transpose = b.T
print("-------------------------------")
print("array b:\n", b)  # Output: [[1 0] [0 1] [1 -1]]
print("Transpose of array b:\n", b_transpose)  # Output: [[1 0 1] [0 1 -1]]
```

```python
# matrix multiplication of a and b
result_ab = np.dot(a, b)  
print("Matrix multiplication of a and b:\n", result_ab)  # Output: [[4 1] [10 1]]

# matrix multiplication of a_transpose and b_transpose
result_ab_transpose = np.dot(a_transpose, b_transpose)
print("Matrix multiplication of a_transpose and b_transpose:\n", result_ab_transpose)  # Output: [[4 1] [10 1]]
```

```python
# check if (a·b)ᵀ = bᵀ·aᵀ using the property: (A·B)ᵀ = Bᵀ·Aᵀ
result_b_transpose_a_transpose = np.dot(b_transpose, a_transpose)
print("Matrix multiplication of b_transpose and a_transpose:\n", result_b_transpose_a_transpose)

# Transpose of a·b
result_ab_transpose = result_ab.T
print("Transpose of the dot product of a and b:\n", result_ab_transpose)

# check if (a·b)ᵀ equals bᵀ·aᵀ using np.allclose() function
is_equal = np.allclose(result_ab_transpose, result_b_transpose_a_transpose)    
print("Is (a·b)ᵀ equal to b_transpose·a_transpose?", is_equal)  # Output: True
```

### Exercise 2 – Ranks of small matrices
1. For each matrix below, compute its rank using np.linalg.matrix_rank:

```python
m1 = np.array([[1, 2], [2, 4]])
m2 = np.array([[1 ,2 ,3], [0, 1, 2], [0, 0, 0]])
m3 = np.array([[1, 0, 1], [0, 1, 1], [1, 1, 2]])

print("Matrix m1:\n", m1)  # Output: [[1 2] [2 4]]
print("--------------------------------")
print("Matrix m2:\n", m2)  # Output: [[1 2 3] [0 1 2] [0 0 0]]
print("--------------------------------")
print("Matrix m3:\n", m3)  # Output: [[1 0 1] [0 1 1] [1 1 2]]
```

```python
# compute the rank of the matrices using np.linalg.matrix_rank() function
rank_m1 = np.linalg.matrix_rank(m1)
rank_m2 = np.linalg.matrix_rank(m2)
rank_m3 = np.linalg.matrix_rank(m3)

print("Rank of matrix m1:", rank_m1)
print("Rank of matrix m2:", rank_m2)
print("Rank of matrix m3:", rank_m3)
```

### Exercise 3 – Rank and null space dimension

```python
n = np.array([[1, 2, 3, 4], [2, 4, 6, 8], [1, 1, 1, 1]])
print("Matrix n:\n", n)  # Output: [[1 2 3 4] [2 4 6 8] [1 1 1 1]]
```

```python
# compute the rank of the matrice n
rank_n = np.linalg.matrix_rank(n)
print("Rank of matrix n:", rank_n)
```

```python
# compute the number of columns of matrix n
num_columns_n = n.shape[1]
print("Number of columns in matrix n:", num_columns_n)
```

```python
# compute the dimension of the null space of matrix n
null_space_dimension_n = n.shape[1] - rank_n
print("Dimension of the null space of matrix n:", null_space_dimension_n)
```

```python
p = np.array([[1, 2, 3], [2, 4, 6], [3, 6, 9]])
print("Matrix p:\n", p)  # Output: [[1 2 3] [2 4 6] [3 6 9]]   
```

```python
# Compute u, s, vh = np.linalg.svd(P)
u, s, vh = np.linalg.svd(p)
print("Matrix u:\n", u)
```

## NumPy: Inverses & Determinants	
Solve linear systems using  NumPy np.linalg.det, np.linalg.inv

```python
# find the inverse of a 2x2 matrix 
# exercise 1

# imnport the Fraction class from the fractions module to display the results as fractions
from fractions import Fraction
np.set_printoptions(
    formatter={'all': lambda x: str(Fraction(x).limit_denominator())}
)

# create a 2x2 matrix
matrix_2x2 = np.array([[-9, -6], [0, 2]])
# compute the determinant of the 2x2 matrix
det_matrix_2x2 = np.linalg.det(matrix_2x2)
print("Determinant of the 2x2 matrix:", det_matrix_2x2)
# check if the determinant is not zero before computing the inverse
if det_matrix_2x2 != 0:
    inverse_matrix_2x2 = np.linalg.inv(matrix_2x2)
    print("Inverse of the 2x2 matrix:\n", inverse_matrix_2x2)
else:
    print("The 2x2 matrix is singular and does not have an inverse.")
```
