### Python Mini-Project: Least Squares Line Fit

Implement linear regression from scratch

```python
# least squares using python
import numpy as np
import matplotlib.pyplot as plt


# generate some data
x = np.array([0, 1, 2, 3, 4], dtype=float)
y = np.array([1, 2, 2, 4, 3], dtype=float)

#  build the matrix A and vector b for the least squares problem
A = np.vstack([x, np.ones(len(x))]).T
b = y

# solve for the coefficients using least squares - solution to Ax = b using the normal equations
coefficients = np.linalg.inv(A.T @ A) @ A.T @ b
slope, intercept = coefficients
print(f"Slope: {slope}, Intercept: {intercept}")
```

```python
# solve for the coefficients using least squares - solution to Ax = b using numpy's lstsq function
coefficients_lstsq, residuals, rank, singular_values = np.linalg.lstsq(A, b, rcond=None)
slope_lstsq, intercept_lstsq = coefficients_lstsq
print(f"Slope (lstsq): {slope_lstsq}, Intercept (lstsq): {intercept_lstsq}")
```

```python
# plot the data and the fitted line
plt.scatter(x, y, color='blue', label='Data Points')
plt.plot(x, slope * x + intercept, color='red', label='Fitted Line (Normal Equations)')
plt.plot(x, slope_lstsq * x + intercept_lstsq, color='green', linestyle='--', label='Fitted Line (lstsq)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression')
plt.legend()
plt.show()
```
