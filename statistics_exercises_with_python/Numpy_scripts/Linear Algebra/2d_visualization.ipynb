## Python Mini-Project: 2D Visualization

```python
# Stage 1: plot a basic shape such as a square or triangle in 2D

import numpy as np
import matplotlib.pyplot as plt
# Define the vertices of a square
square_vertices = np.array([[0, 0], [1, 0], [1, 1], [0, 1], [0, 0]])  # Closing the square by repeating the first vertex
# Plot the square  
plt.plot(square_vertices[:, 0], square_vertices[:, 1], marker='o')
plt.title('Square in 2D')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis('equal')
plt.show()
```

```python
# Stage 2: apply one transformation matrix, like scaling, rotation, shear, or reflection
# Define a scaling transformation matrix
scaling_matrix = np.array([[2, 0], [0, 2]])
# Apply the scaling transformation to the square vertices
scaled_square = square_vertices @ scaling_matrix.T
# Plot the original and scaled squares
plt.plot(square_vertices[:, 0], square_vertices[:, 1], marker='o', label='Original Square')
plt.plot(scaled_square[:, 0], scaled_square[:, 1], marker='o', label='Scaled Square')
plt.title('Scaling Transformation')
plt.xlabel('X')
plt.ylabel('Y') 
plt.axis('equal')
plt.legend()
plt.show()
```

```python
# stage 3: apply multiple transformations in sequence, such as a rotation followed by a scaling, and visualize the result
# Define a rotation transformation matrix (45 degrees)
rotation_matrix = np.array([[np.cos(np.pi/4), -np.sin(np.pi/4)], [np.sin(np.pi/4), np.cos(np.pi/4)]])
# Apply the rotation transformation to the square vertices
rotated_square = square_vertices @ rotation_matrix.T
# Define a scaling transformation matrix
scaling_matrix = np.array([[2, 0], [0, 2]])
# Apply the scaling transformation to the rotated square vertices
transformed_square = rotated_square @ scaling_matrix.T
# Plot the original, rotated, and transformed squares
plt.plot(square_vertices[:, 0], square_vertices[:, 1], marker='o', label='Original Square')
plt.plot(rotated_square[:, 0], rotated_square[:, 1], marker='o', label='Rotated Square')
plt.plot(transformed_square[:, 0], transformed_square[:, 1], marker='o', label='Transformed Square')
plt.title('Multiple Transformations')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis('equal')
plt.legend()
plt.show()
```
